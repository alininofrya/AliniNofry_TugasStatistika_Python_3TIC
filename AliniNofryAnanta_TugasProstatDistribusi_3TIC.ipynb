```python
ruang_sampel = [1, 2, 3, 4, 5, 6]

# Menentukan perangkat dengan nomor genap
kejadian = [x for x in ruang_sampel if x % 2 == 0]

# Menghitung probabilitas
probabilitas = len(kejadian) / len(ruang_sampel)

print("Ruang Sampel:", ruang_sampel)
print("Kejadian yang Memenuhi:", kejadian)
print("Probabilitas:", probabilitas)
print("Persentase:", probabilitas * 100, "%")
```

```python
probabilitas_berhasil = 0.95

# Menghitung probabilitas kegagalan
probabilitas_gagal = 1 - probabilitas_berhasil

print("Probabilitas Berhasil:", probabilitas_berhasil)
print("Probabilitas Gagal:", probabilitas_gagal)
print("Persentase Kegagalan:", probabilitas_gagal * 100, "%")
```

```python
total_perangkat = 100
windows = 40
linux = 30
keduanya = 10

prob_windows = windows / total_perangkat
prob_linux = linux / total_perangkat
prob_keduanya = keduanya / total_perangkat

# Menghitung probabilitas gabungan
prob_gabungan = prob_windows + prob_linux - prob_keduanya

print("Probabilitas Windows:", prob_windows)
print("Probabilitas Linux:", prob_linux)
print("Probabilitas Keduanya:", prob_keduanya)
print("Probabilitas Windows atau Linux:", prob_gabungan)
print("Persentase:", prob_gabungan * 100, "%")
```

```python
total_pengguna = 100
pengguna_android = 60
android_dengan_fitur = 30

# Menghitung probabilitas bersyarat
probabilitas = android_dengan_fitur / pengguna_android

print("Total Pengguna:", total_pengguna)
print("Pengguna Android:", pengguna_android)
print("Android dengan Fitur Keamanan:", android_dengan_fitur)
print("Probabilitas Bersyarat:", probabilitas)
print("Persentase:", probabilitas * 100, "%")
```

```python
prob_serangan = 0.05
prob_deteksi = 0.90
prob_peringatan_palsu = 0.10

# Menghitung probabilitas tidak terjadi serangan
prob_tidak_serangan = 1 - prob_serangan

# Menghitung probabilitas munculnya peringatan
prob_peringatan = (
    prob_deteksi * prob_serangan
    + prob_peringatan_palsu * prob_tidak_serangan
)

# Menghitung Teorema Bayes
prob_serangan_setelah_peringatan = (
    prob_deteksi * prob_serangan / prob_peringatan
)

print("Probabilitas Peringatan:", prob_peringatan)
print("Probabilitas Serangan setelah Peringatan:",
      prob_serangan_setelah_peringatan)
print("Persentase:",
      prob_serangan_setelah_peringatan * 100, "%")
```

```python
import random

random.seed(42)

jumlah_percobaan = 10000
jumlah_angka_6 = 0

for i in range(jumlah_percobaan):
    hasil = random.randint(1, 6)

    if hasil == 6:
        jumlah_angka_6 += 1

# Menghitung probabilitas simulasi
probabilitas_simulasi = jumlah_angka_6 / jumlah_percobaan

# Menghitung probabilitas teoritis
probabilitas_teoritis = 1 / 6

print("Jumlah Percobaan:", jumlah_percobaan)
print("Kemunculan Angka 6:", jumlah_angka_6)
print("Probabilitas Simulasi:", probabilitas_simulasi)
print("Probabilitas Teoritis:", probabilitas_teoritis)
```

```python
import random

random.seed(42)

jumlah_percobaan = 10000
probabilitas_berhasil = 0.95

berhasil = 0

for i in range(jumlah_percobaan):
    hasil = random.random()

    if hasil < probabilitas_berhasil:
        berhasil += 1

# Menghitung probabilitas simulasi
prob_simulasi = berhasil / jumlah_percobaan

print("Jumlah Percobaan:", jumlah_percobaan)
print("Jumlah Pengiriman Berhasil:", berhasil)
print("Probabilitas Teoritis:", probabilitas_berhasil)
print("Probabilitas Simulasi:", prob_simulasi)
print("Selisih:", abs(probabilitas_berhasil - prob_simulasi))
```

```python
from scipy.stats import bernoulli
import numpy as np
import matplotlib.pyplot as plt

# Parameter distribusi
p = 0.8

# Menghitung probabilitas
prob_berhasil = bernoulli.pmf(1, p)
prob_gagal = bernoulli.pmf(0, p)

# Menghitung nilai harapan dan varians
nilai_harapan = bernoulli.mean(p)
varians = bernoulli.var(p)

print("Probabilitas Berhasil:", prob_berhasil)
print("Probabilitas Gagal:", prob_gagal)
print("Nilai Harapan:", nilai_harapan)
print("Varians:", varians)

# Visualisasi distribusi
x = np.array([0, 1])
y = bernoulli.pmf(x, p)

plt.bar(x, y)
plt.title("Distribusi Bernoulli")
plt.xlabel("Hasil Login (0 = Gagal, 1 = Berhasil)")
plt.ylabel("Probabilitas")
plt.xticks([0, 1])
plt.ylim(0, 1)
plt.show()
```

```python
from scipy.stats import binom
import numpy as np
import matplotlib.pyplot as plt

# Parameter distribusi
n = 5
p = 0.8
x = 4

# Menghitung probabilitas
probabilitas = binom.pmf(x, n, p)

# Menghitung nilai harapan dan varians
nilai_harapan = binom.mean(n, p)
varians = binom.var(n, p)

print("Probabilitas Tepat 4 Berhasil:", probabilitas)
print("Nilai Harapan:", nilai_harapan)
print("Varians:", varians)

# Visualisasi distribusi
nilai_x = np.arange(0, n + 1)
nilai_y = binom.pmf(nilai_x, n, p)

plt.bar(nilai_x, nilai_y)
plt.title("Distribusi Binomial")
plt.xlabel("Jumlah Pengiriman Berhasil")
plt.ylabel("Probabilitas")
plt.xticks(nilai_x)
plt.show()
```
